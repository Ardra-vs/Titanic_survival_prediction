filling missing values

```python
df['Age']=df['Age'].fillna(df['Age'].median())
```

```python
df['Embarked']=df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df
```

Encoding

```python
dfe=pd.get_dummies(df[["Sex","Embarked"]],drop_first=True,dtype=int)
dfe
```

```python
dfN=pd.concat([df,dfe],axis=1)
```

```python
dfN.drop(columns=["Sex","Embarked"],axis=1,inplace=True)
```

```python
dfN.columns
```

```python
#x_y splitting
x=dfN.drop(columns='Survived')
x.ndim
```

```python
y=dfN['Survived']
y.ndim
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
x_train.shape
```

```python
y_test.shape
```

```python
# solving imbalance
from imblearn.over_sampling import SMOTE
smote=SMOTE(random_state=42)
x_train_res,y_train_res=smote.fit_resample(x_train,y_train)
```

```python
x_train_res
```

```python
print(y_train.value_counts())
```

```python
print(y_train_res.value_counts())
```

```python
#model training and comparing each model
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train_res)
x_test_scaled=scaler.transform(x_test)
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
knn=KNeighborsClassifier()
nb=GaussianNB()
dt=DecisionTreeClassifier()
lr=LogisticRegression()
```

```python
lst=[knn,nb,dt,lr]
for i in lst:
  print('model is',i)
  if i in (knn,lr):
    i.fit(x_train_scaled,y_train_res)
    y_pred=i.predict(x_test_scaled)
  else:
    i.fit(x_train_res,y_train_res)
    y_pred=i.predict(x_test)
  print('evaluation')

  from sklearn.metrics import accuracy_score,classification_report
  report=classification_report(y_test,y_pred)
  print(report)

  from sklearn.metrics import confusion_matrix
  cm=confusion_matrix(y_test,y_pred)
  print(cm)

  from sklearn.metrics import ConfusionMatrixDisplay
  cmd=ConfusionMatrixDisplay(cm)
  cmd.plot()

  score=accuracy_score(y_test,y_pred)
  print(score)
  print('-'*100)
```

```python
#cress validate the model
from sklearn.model_selection import GridSearchCV
model1=LogisticRegression(max_iter=1000)
data= {'C': [0.01, 0.1, 1, 10],
       'penalty': ['l2'],
       'solver': ['lbfgs'],
       'class_weight': [None, 'balanced']}
clf=GridSearchCV(model1,data,cv=10,scoring='accuracy')
clf.fit(x_train_scaled,y_train_res)
```

```python
print(clf.best_params_)
```

```python
print(clf.best_score_)
```
